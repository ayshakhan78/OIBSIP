```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df= sns.load_dataset("iris")
```

```python
from sklearn.model_selection import train_test_split
X=df.drop("species",axis=1)
y=df["species"]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model_knn=KNeighborsClassifier(n_neighbors=13)
```

```python
model_knn.fit(X_train,y_train)
```

```python
model_knn.score(X_test,y_test)
```

```python
from sklearn.svm import SVC
model_svm=SVC(gamma="auto")
model_svm.fit(X_train,y_train)
```

```python
model_svm.score(X_test,y_test)
```

```python
from sklearn.model_selection import GridSearchCV
classifier_svm = GridSearchCV(model_svm,{
    "C" : [10,20,30],
    "kernel" :["linear","rbf"]
},cv=5,return_train_score=False)
```

```python
classifier_svm.fit(X,y)
```

```python
classifier_svm.cv_results_
```

```python
results = pd.DataFrame(classifier_svm.cv_results_)
```

```python
results
```

```python
results[["param_C","param_kernel","mean_test_score"]].sort_values(by="mean_test_score",ascending=False)
```

```python
classifier_knn=GridSearchCV(model_knn,{
    "n_neighbors" : [5,7,9,11,13],
    "weights" : ["uniform","distance"],
    "algorithm" : ["auto", "ball_tree", "kd_tree", "brute"]
},cv=5,return_train_score=False)
```

```python
classifier_knn.fit(X  ,y)
```

```python
classifier_knn.cv_results_
```
